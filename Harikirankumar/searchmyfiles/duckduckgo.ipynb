```python
from ddgs import DDGS
from g4f.client import Client

# ── Step 1: Search and grab the top page URL
query = "Canadian T1 income tax return guide CRA"
results = DDGS().text(query, region="ca-en", max_results=3)
top_url = results[0]["href"]
print(f"Extracting content from: {top_url}\n")

# ── Step 2: Extract full page text using ddgs.extract()
# Note: extract() takes a single URL string, not a list
page = DDGS().extract(top_url)          # returns a dict with 'title', 'text', etc.
page_text = page.get("text", "") or page.get("body", "")
# Trim to ~3000 chars to stay within context limits of free providers
content_snippet = page_text[:3000].strip()

print(f"Page title : {page.get('title', '')}")
print(f"Content length: {len(page_text)} chars — sending first 3000 to AI...\n")

# ── Step 3: Ask g4f (free, no API key) to summarize the content
client = Client()

prompt = f"""You are a helpful tax assistant.
Below is content extracted from a CRA / tax guide webpage.
Give a clear, structured AI-style summary covering:
- What a T1 return is
- Who needs to file
- Key deadlines
- How to file (online/paper)
- Common deductions and credits

PAGE CONTENT:
{content_snippet}

Provide a well-formatted summary."""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}],
)

print("=" * 60)
print("AI Summary — Canadian T1 Return")
print("=" * 60)
print(response.choices[0].message.content)
```

```python
from ddgs import DDGS
from urllib.parse import urlparse
import re


def run_query(q, region="ca-en", max_results=5):
    rows = list(DDGS().text(q, region=region, safesearch="moderate", max_results=max_results))
    cleaned = []
    for row in rows:
        cleaned.append({
            "title": row.get("title", ""),
            "href": row.get("href", ""),
            "body": (row.get("body", "") or "").replace("\n", " "),
        })
    return cleaned


def score_results(rows, required_terms=(), preferred_domains=()):
    score = 0
    for idx, row in enumerate(rows[:5], start=1):
        text = f"{row['title']} {row['body']} {row['href']}".lower()
        domain = urlparse(row["href"]).netloc.lower()
        weight = max(1, 6 - idx)
        term_hits = sum(1 for term in required_terms if term.lower() in text)
        domain_hit = any(d.lower() in domain for d in preferred_domains)
        score += term_hits * weight * 3
        if domain_hit:
            score += weight * 4
    return score


scenarios = [
    {
        "name": "Greeting-polluted CRA T1 question",
        "required_terms": ["t1", "cra", "income tax", "return"],
        "preferred_domains": ["canada.ca", "cra", "intuit.ca"],
        "queries": [
            "Hello What do you know about canadian pdf t1 returns",
            "what do you know about canadian pdf t1 returns",
            "canadian t1 return pdf cra",
            "CRA T1 General income tax and benefit return pdf",
            "CRA T1 return guide canada",
        ],
    },
    {
        "name": "Hugging Face explainer",
        "required_terms": ["hugging face", "models", "ai"],
        "preferred_domains": ["huggingface.co", "wikipedia.org"],
        "queries": [
            "What is hugging face what do you know about it?",
            "hugging face ai platform",
            "Hugging Face overview models datasets",
        ],
    },
    {
        "name": "Microsoft Phi model info",
        "required_terms": ["phi", "microsoft", "model"],
        "preferred_domains": ["microsoft.com", "huggingface.co"],
        "queries": [
            "hey can you explain phi-3 model",
            "Microsoft Phi-3 model overview",
            "microsoft phi 3 mini instruct hugging face",
        ],
    },
    {
        "name": "Canadian GST/HST filing",
        "required_terms": ["gst", "hst", "return", "cra"],
        "preferred_domains": ["canada.ca"],
        "queries": [
            "what is GST HST return filing in canada",
            "CRA GST HST return filing guide",
            "site:canada.ca GST HST return CRA",
        ],
    },
    {
        "name": "PDF-specific government form lookup",
        "required_terms": ["pdf", "form", "canada"],
        "preferred_domains": ["canada.ca"],
        "queries": [
            "site:canada.ca T1 return pdf",
            "canada t1 form pdf",
            "CRA income tax package pdf canada",
        ],
    },
]

all_results = []
for scenario in scenarios:
    print("=" * 110)
    print("SCENARIO:", scenario["name"])
    ranked = []
    for q in scenario["queries"]:
        try:
            rows = run_query(q)
            score = score_results(rows, scenario["required_terms"], scenario["preferred_domains"])
            ranked.append((score, q, rows))
            print(f"\nQUERY: {q}\nSCORE: {score}")
            for i, row in enumerate(rows[:3], start=1):
                print(f"  {i}. {row['title']}")
                print(f"     {row['href']}")
                print(f"     {row['body'][:150]}")
        except Exception as e:
            print(f"\nQUERY: {q}\nERROR: {e!r}")
    ranked.sort(reverse=True, key=lambda x: x[0])
    if ranked:
        best_score, best_query, _ = ranked[0]
        all_results.append((scenario["name"], best_query, best_score))
        print(f"\nBEST QUERY => {best_query}  (score={best_score})")
    print()

print("#" * 110)
print("BEST QUERIES SUMMARY")
for name, best_query, best_score in all_results:
    print(f"- {name}: {best_query} (score={best_score})")
```

```python
from ddgs import DDGS
from urllib.parse import urlparse
import re
from collections import defaultdict


def normalize_user_query(text: str) -> str:
    q = str(text or "").strip()
    q = re.sub(r"\s+", " ", q)
    q = re.sub(r"^(hi|hello|hey)\b[\s,!.:-]*", "", q, flags=re.I)
    q = re.sub(r"\b(what do you know about|can you explain|tell me about|i need help with|please help with)\b", "", q, flags=re.I)
    q = re.sub(r"\s+", " ", q).strip(" ?.-")
    return q


def query_candidates(raw_query: str, official_domain: str | None = None, official_hint: str | None = None):
    base = normalize_user_query(raw_query)
    candidates = [raw_query, base]
    if official_hint:
        candidates.append(official_hint)
        if base:
            candidates.append(f"{official_hint} {base}")
    if official_domain and base:
        candidates.append(f"site:{official_domain} {base}")
    # de-duplicate while preserving order
    out = []
    seen = set()
    for q in candidates:
        q = re.sub(r"\s+", " ", str(q or "")).strip()
        if q and q not in seen:
            seen.add(q)
            out.append(q)
    return out


def fetch_rows(q: str, region: str = "wt-wt", max_results: int = 5):
    rows = list(DDGS().text(q, region=region, safesearch="moderate", max_results=max_results))
    return [
        {
            "title": str(r.get("title", "")),
            "href": str(r.get("href", "")),
            "body": str(r.get("body", "") or "").replace("\n", " "),
        }
        for r in rows
    ]


def score_rows(rows, required_terms=(), preferred_domains=()):
    score = 0
    reasons = []
    for idx, row in enumerate(rows[:5], start=1):
        text = f"{row['title']} {row['body']} {row['href']}".lower()
        domain = urlparse(row["href"]).netloc.lower()
        weight = max(1, 6 - idx)
        hits = [t for t in required_terms if t.lower() in text]
        if hits:
            score += len(hits) * weight * 3
            reasons.append(f"rank{idx}:terms={','.join(hits[:3])}")
        if any(d.lower() in domain for d in preferred_domains):
            score += weight * 5
            reasons.append(f"rank{idx}:domain={domain}")
        if row["href"].startswith("https://"):
            score += 0.5
    return score, reasons


scenarios = [
    {
        "name": "Canadian tax / CRA",
        "raw_query": "Hello what do you know about canadian pdf t1 returns",
        "required_terms": ["t1", "cra", "return", "income tax"],
        "preferred_domains": ["canada.ca", "cra"],
        "official_domain": "canada.ca",
        "official_hint": "CRA T1 General income tax and benefit return pdf",
        "region": "ca-en",
    },
    {
        "name": "AI platform / Hugging Face",
        "raw_query": "hey what do you know about Hugging Face",
        "required_terms": ["hugging face", "models", "datasets", "spaces"],
        "preferred_domains": ["huggingface.co", "wikipedia.org"],
        "official_domain": "huggingface.co",
        "official_hint": "Hugging Face overview models datasets spaces",
        "region": "us-en",
    },
    {
        "name": "Microsoft model docs",
        "raw_query": "can you explain phi-3 mini",
        "required_terms": ["phi", "microsoft", "model", "mini"],
        "preferred_domains": ["microsoft.com", "huggingface.co", "azure.com"],
        "official_domain": "microsoft.com",
        "official_hint": "Microsoft Phi-3 mini model",
        "region": "us-en",
    },
    {
        "name": "Government service / passport",
        "raw_query": "I need help with canada passport renewal",
        "required_terms": ["passport", "renewal", "canada"],
        "preferred_domains": ["canada.ca"],
        "official_domain": "canada.ca",
        "official_hint": "Canada passport renewal",
        "region": "ca-en",
    },
    {
        "name": "Developer docs / FastAPI",
        "raw_query": "what do you know about fastapi dependency injection",
        "required_terms": ["fastapi", "dependency", "dependencies"],
        "preferred_domains": ["fastapi.tiangolo.com", "github.com"],
        "official_domain": "fastapi.tiangolo.com",
        "official_hint": "FastAPI dependencies tutorial",
        "region": "us-en",
    },
    {
        "name": "Cloud service / Azure OpenAI",
        "raw_query": "tell me about azure openai pricing",
        "required_terms": ["azure", "openai", "pricing"],
        "preferred_domains": ["azure.microsoft.com", "learn.microsoft.com"],
        "official_domain": "learn.microsoft.com",
        "official_hint": "Azure OpenAI pricing",
        "region": "us-en",
    },
    {
        "name": "Troubleshooting / Python SSL error",
        "raw_query": "hello how to fix python ssl certificate verify failed",
        "required_terms": ["python", "ssl", "certificate", "verify failed"],
        "preferred_domains": ["docs.python.org", "stackoverflow.com", "github.com"],
        "official_domain": "docs.python.org",
        "official_hint": "Python SSL certificate verify failed",
        "region": "us-en",
    },
]

summary = []
for scenario in scenarios:
    print("=" * 120)
    print("SCENARIO:", scenario["name"])
    candidates = query_candidates(
        scenario["raw_query"],
        official_domain=scenario.get("official_domain"),
        official_hint=scenario.get("official_hint"),
    )
    ranked = []
    for q in candidates:
        try:
            rows = fetch_rows(q, region=scenario.get("region", "wt-wt"), max_results=5)
            score, reasons = score_rows(rows, scenario["required_terms"], scenario["preferred_domains"])
            ranked.append({"query": q, "score": score, "reasons": reasons, "rows": rows})
            print(f"\nQUERY: {q}\nSCORE: {score:.1f}")
            for i, row in enumerate(rows[:3], start=1):
                print(f"  {i}. {row['title']}")
                print(f"     {row['href']}")
                print(f"     {row['body'][:140]}")
        except Exception as e:
            print(f"\nQUERY: {q}\nERROR: {e!r}")
    ranked.sort(key=lambda x: x["score"], reverse=True)
    if ranked:
        best = ranked[0]
        summary.append((scenario["name"], best["query"], best["score"], best["reasons"][:4]))
        print("\nBEST QUERY:", best["query"])
        print("WHY:", "; ".join(best["reasons"][:6]) or "best aggregate relevance")

print("\n" + "#" * 120)
print("BEST QUERY SUMMARY")
for name, query, score, reasons in summary:
    print(f"- {name}: {query} | score={score:.1f}")
    if reasons:
        print(f"  reasons: {'; '.join(reasons)}")
```

```python
# Real query dataset sampling benchmark (general-purpose)
# Uses public query datasets from Hugging Face to test DDG behavior beyond handcrafted prompts.

from ddgs import DDGS
import random
import re

try:
    from datasets import load_dataset
except Exception:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "datasets", "-q"])
    from datasets import load_dataset


def sanitize_query(q: str) -> str:
    q = str(q or "").strip()
    q = re.sub(r"\s+", " ", q)
    q = q.strip(" \n\t\r.,;:!?\"'()[]{}")
    return q[:160]


def pick_queries(dataset_name, split_name, field, n=15, seed=42, config_name=None):
    if config_name:
        ds = load_dataset(dataset_name, config_name, split=split_name)
    else:
        ds = load_dataset(dataset_name, split=split_name)
    vals = []
    for row in ds:
        text = sanitize_query(row.get(field, ""))
        if len(text) >= 8:
            vals.append(text)
    random.seed(seed)
    random.shuffle(vals)
    uniq = []
    seen = set()
    for v in vals:
        key = v.lower()
        if key not in seen:
            seen.add(key)
            uniq.append(v)
        if len(uniq) >= n:
            break
    return uniq


def ddg_probe(query, region="wt-wt", max_results=5):
    rows = list(DDGS().text(query, region=region, safesearch="moderate", max_results=max_results))
    cleaned = []
    for row in rows[:max_results]:
        cleaned.append({
            "title": str(row.get("title", "")),
            "href": str(row.get("href", "")),
            "body": str(row.get("body", "") or "").replace("\n", " "),
        })
    return cleaned


def score_generic(rows, query):
    q_terms = [t for t in re.findall(r"[a-zA-Z0-9]+", query.lower()) if len(t) >= 4][:6]
    if not rows:
        return 0.0
    score = 0.0
    for i, row in enumerate(rows, start=1):
        text = f"{row['title']} {row['body']}".lower()
        overlap = sum(1 for t in q_terms if t in text)
        weight = max(1, 6 - i)
        score += overlap * weight
        if row["href"].startswith("https://"):
            score += 0.25
    return score


benchsets = [
    # MS MARCO real web questions
    {"name": "MS MARCO v1.1 train", "dataset": "microsoft/ms_marco", "config": "v1.1", "split": "train", "field": "query", "region": "us-en", "n": 12},
    # BEIR FiQA financial retrieval queries
    {"name": "BEIR FiQA queries", "dataset": "BeIR/fiqa", "config": "queries", "split": "queries", "field": "text", "region": "us-en", "n": 12},
    # BEIR SciFact scientific claim retrieval queries
    {"name": "BEIR SciFact queries", "dataset": "BeIR/scifact", "config": "queries", "split": "queries", "field": "text", "region": "wt-wt", "n": 12},
]

overall = []
for cfg in benchsets:
    print("=" * 120)
    print("DATASET:", cfg["name"])
    queries = pick_queries(
        cfg["dataset"],
        cfg["split"],
        cfg["field"],
        n=cfg["n"],
        seed=42,
        config_name=cfg.get("config"),
    )
    ds_scores = []
    for idx, q in enumerate(queries, start=1):
        try:
            rows = ddg_probe(q, region=cfg["region"], max_results=5)
            s = score_generic(rows, q)
            ds_scores.append(s)
            top = rows[0]["title"] if rows else "<no result>"
            print(f"{idx:02d}. score={s:5.1f} | q={q[:90]}")
            print(f"    top: {top}")
        except Exception as e:
            print(f"{idx:02d}. ERROR | q={q[:90]} | {e!r}")
    if ds_scores:
        avg = sum(ds_scores) / len(ds_scores)
        med = sorted(ds_scores)[len(ds_scores) // 2]
        overall.append((cfg["name"], avg, med, len(ds_scores)))
        print(f"\nSUMMARY {cfg['name']}: avg={avg:.2f}, median={med:.2f}, n={len(ds_scores)}")

print("\n" + "#" * 120)
print("OVERALL DATASET SUMMARY")
for name, avg, med, n in overall:
    print(f"- {name}: avg={avg:.2f}, median={med:.2f}, n={n}")
```
